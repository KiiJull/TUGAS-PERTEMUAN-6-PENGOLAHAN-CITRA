```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Baca citra
img = cv2.imread('WhatsApp Image 2025-10-24 at 16.56.12 (2).jpeg')   # ganti nama sesuai file kamu
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Tampilkan citra
plt.imshow(img_rgb)
plt.title("Citra Asli (Kijul)")
plt.axis('off')
plt.show()

# Hitung ukuran dan aspect ratio
height, width, channel = img.shape
print("Ukuran Citra:", width, "x", height)
print("Resolusi (jumlah piksel):", width * height)
print("Aspect Ratio:", round(width / height, 2))
```
